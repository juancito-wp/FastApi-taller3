# 👁️🤖 ¿Qué es la visión artificial?

La **visión artificial** es una tecnología que permite a las computadoras 🖥️ **captar, analizar e interpretar imágenes y videos** 📸🎥, de una manera similar a como los seres humanos 👀 interpretan lo que ven.

Esta tecnología utiliza **cámaras 📷, sensores 🔎, inteligencia artificial 🧠 y algoritmos ⚙️** para identificar objetos 🚗, personas 👤, formas 🔵, colores 🎨, movimientos 🏃 y patrones dentro de una imagen.

## ⚙️ ¿Cómo funciona?

La visión artificial generalmente sigue estos pasos:

1. 📸 **Captura de imágenes:** una cámara o sensor obtiene una imagen o video.
2. 🖥️ **Procesamiento:** el sistema mejora y prepara la imagen para analizarla.
3. 🔍 **Análisis:** los algoritmos identifican características, objetos o patrones.
4. 🧠 **Interpretación:** la computadora determina qué está viendo y puede tomar una decisión según la información obtenida.

## 🌎 Ejemplos de uso

- 🚗 **Vehículos autónomos:** detectan otros vehículos, peatones 🚶 y señales de tránsito 🚦.
- 🏥 **Medicina:** ayuda a detectar enfermedades mediante imágenes médicas 🩻.
- 🏭 **Industria:** permite revisar productos 📦 y detectar errores o defectos ❌.
- 📱 **Reconocimiento facial:** identifica personas mediante sus características faciales 👤.
- 🛒 **Comercio:** ayuda a controlar inventarios 📊 y analizar productos.
- 📹 **Seguridad:** permite detectar movimientos o situaciones específicas mediante cámaras 🔐.

## ⭐ Importancia

La visión artificial permite **automatizar tareas que requieren analizar imágenes** 🤖📸, aumentando la rapidez ⚡ y precisión 🎯 de diferentes procesos.

Actualmente, es una de las aplicaciones más importantes de la **inteligencia artificial 🧠🤖** y se utiliza en áreas como la medicina 🏥, la industria 🏭, el transporte 🚘 y la seguridad 🔒.

## 👤📸 Detección de rostros con Python y OpenCV

La **detección de rostros** es una técnica de visión artificial 🤖👁️ que permite identificar caras humanas en imágenes 📷 o videos 🎥. Con **Python 🐍 y OpenCV**, es posible detectar rostros en tiempo real y marcarlos automáticamente 🟩.

```python
#instalacion de open cv

#python -m pip install opencv-python

# importcion de la libreria
import cv2

#cargar la imagen
imgpath = 'faces.webp'

#Leer la imagen
img = cv2.imread(imgpath)

#Mostrar la imagen
img.shape

print("Version", cv2.__version__)
print("CascadeClassifier", hasattr(cv2, 'CascadeClassifier'))

```

```python
#Convertir la imagen a escala de grises
gray_img =cv2.cvtColor(img, cv2.COLOR_BGR2GRAY)
cv2.imshow('Gray Image', gray_img)
```

```python
cascade = "haarcascade_frontalface_default.xml"
face_cascade = cv2.CascadeClassifier(cascade)

if face_cascade.empty():
    print("error: no se pudo cargar, verf¿ifica el xml")
else:
    print("cargado correctamente")
```

```python
face = face_cascade.detectMultiScale(gray_img, scaleFactor=1.1, minNeighbors=5, minSize=(40, 40))
```

```python
for (x, y, w, h) in face:
    cv2.rectangle(img, (x, y), (x + w, y + h), (0, 255, 0), 2)
    
img_rgb = cv2.cvtColor(img, cv2.COLOR_BGR2RGB)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(20, 10))
plt.imshow(img_rgb)
plt.axis('off')
```

Captura de video en tiempo real


```python
#1. Acceder a la camara de nuestro pc y leer un flujo de datos en tiempo real.

video_capture = cv2.VideoCapture(0)

#2.Funcion para detectar rostro con flujo de video y dibujar un recuadro delimitador

def detect_bounding_box(vid):
    gray_img =  cv2.cvtColor(vid, cv2.COLOR_BGR2GRAY)
    face = face_cascade.detectMultiScale(gray_img, 1.1, 5, minSize=(40, 40))
    for (x, y, w, h) in face:
        cv2.rectangle(vid, (x, y), (x + w, y + h), (0, 255, 0), 2)
    return face

    
```

```python
#3. While para detectar caras en Real Time, capturar fotograma y aplicar deteccion de rostros

while True:
    result, video_frame = video_capture.read()

    if result is False:
        break
    faces = detect_bounding_box(video_frame)

    cv2.imshow("Mi cara detectada", video_frame)
    if cv2.waitKey(1) & 0xFF == ord("q"):
        break

video_capture.release()
cv2.destroyAllWindows()
```
